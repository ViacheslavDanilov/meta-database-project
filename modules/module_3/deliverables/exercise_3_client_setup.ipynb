This notebook uses credentials from a `.env` file.

Expected variables: `DB_HOST`, `DB_PORT`, `DB_USER`, `DB_PASSWORD`, `DB_NAME`.

```python
import os
from dotenv import find_dotenv, load_dotenv
import mysql.connector as connector
```

```python
dotenv_path = find_dotenv(usecwd=True)
if not dotenv_path:
    raise FileNotFoundError(
        "No .env file found. Copy .env.example to .env and fill DB credentials."
    )

load_dotenv(dotenv_path)
print(f"Loaded env from: {dotenv_path}")
```

```python
required_vars = ["DB_HOST", "DB_PORT", "DB_USER", "DB_PASSWORD", "DB_NAME"]
missing = [v for v in required_vars if not os.getenv(v)]
if missing:
    raise ValueError(f"Missing env vars: {', '.join(missing)}")

connection = connector.connect(
    host=os.getenv("DB_HOST"),
    port=int(os.getenv("DB_PORT", "3306")),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    database=os.getenv("DB_NAME"),
)

print("Connected:", connection.is_connected())
```

```python
cursor = connection.cursor()
cursor.execute("SHOW TABLES;")
tables = cursor.fetchall()
print("Tables:", tables[:10])
```

```python
cursor.close()
connection.close()
print("Connection closed")
```
